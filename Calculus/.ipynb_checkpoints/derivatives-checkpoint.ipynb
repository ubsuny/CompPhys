## Now check the accuracy of the methods for a few functions

Including the crazy $\sin(1/x)$. 

```python
a = 0.01
b = np.pi
xvals = np.linspace(a,b,10000)
yvals = np.sin(1./xvals)
plt.xlabel("x")
plt.ylabel("sin(1/x)")
plt.plot(xvals,yvals)
plt.show()
```

```python
print("Testing algorithms for computing numeric derivatives")

h = 0.01
x = 0.001

# name: (f(x), analytical f'(x))
funcs = {
    "exp":      (np.exp,                np.exp),
    "sin":      (np.sin,                np.cos),
    "x^2":      (lambda x: x**2,        lambda x: 2*x),
    "sin(1/x)": (lambda x: np.sin(1/x), lambda x: -np.cos(1/x) / x**2),
}

# name: deriv(f, x, h) -> derivative estimate
derivative_algos = {
    "forward":   derivative_forward,
    "backward":  derivative_backward,
    "symmetric": derivative_symmetric,
    "5-point":   derivative_fivepoint,
    "ridders":   lambda f, x, h: derivative_ridders(f, x, h)[0],  # drop error estimate
}

# results[row][column] = value; "exact" row first
results = {"exact": {name: df(x) for name, (f, df) in funcs.items()}}
for algo, deriv in derivative_algos.items():
    results[algo] = {name: deriv(f, x, h) for name, (f, df) in funcs.items()}

def print_table(title, table, fmt, col_w=19, row_w=10):
    print(title)
    print(f"{'':<{row_w}}" + "".join(f"{name:>{col_w}}" for name in funcs))
    print("-" * (row_w + col_w * len(funcs)))
    for row_name, row in table.items():
        print(f"{row_name:<{row_w}}" + "".join(f"{format(row[name], fmt):>{col_w}}" for name in funcs))
    print()

print(f"x = {x},  h = {h}\n")
print_table("f'(x)", results, "+.10e")

rel_err = {algo: {name: abs(row[name] / results["exact"][name] - 1) for name in funcs}
           for algo, row in results.items() if algo != "exact"}
print_table("|relative error|", rel_err, ".1e")
```
